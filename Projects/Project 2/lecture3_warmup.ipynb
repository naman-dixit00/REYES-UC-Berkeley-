Now, create a dictionary to organize the sports into similar categories (i.e. skating sports, ball sports, etc.)

Recall dictionary structure: my_dictionary = {"Key": "Value", "Key2": "Value 2" "Value 3"}

```python
sports_dict = {
    "Ball Sports": ["soccer", "football", "tennis", "pickleball", "volleyball", "baseball"],
    "Water Sports": ["swimming", "waterpolo"],
    "Running Sports": ["running"],
    "Winter Sports": ["hockey"]
}
```

Access and print the first sport of the second category in sports_dict.

```python
print(sports_dict["Water Sports"][0])
```

Use code to identify and confirm the data type of the key that you printed in the previous step.

```python
print(type(sports_dict["Water Sports"][0]))
```

Write a function that identifies the type of a key within a dictionary.

```python
def get_key_type(dictionary, key_name):
    if key_name in dictionary:
        return type(key_name)
    else:
        return f"Key '{key_name}' not found in the dictionary."

# Example usage:
print(get_key_type(sports_dict, "Ball Sports"))
print(get_key_type(sports_dict, "Running Sports"))
print(get_key_type(sports_dict, "NonExistentKey"))
```
