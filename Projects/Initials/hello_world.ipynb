Today we will explore games like Angry Bird.
These are games in which a projectile (a bird, a cannon ball, etc...) are shot with parameters determined by the user.
The trajectory the projectile follows will then be determined by the laws of physics, which we will explore thoroughly in the lectures to come.
The question we want to answer today is whether the trajectory hits a target at some predetermined position.

## User input

Newton: to determine the trajectory of a particle, you need to know its initial position and velocity. By convention, we will set up our system so that its initial position is the point $(x,y) = (0,0)$.

```python
initial_velocity = 10 # measured in meters per second
angle = 30 # measured in degrees from the horizon
```

```python
print(f"The initial velocity was {initial_velocity} meters per second")
print(f"The initial angle was {angle} degrees from the horizon")
```

## Time evolution

Newton's equation say that at time $t$, the position of the projectile will be
$$y(t) = v_0\sin(\theta)t - \frac{1}{2}gt^2$$
$$x(t) = v_0\cos(\theta)t$$
In these equations $v_0$ is the initial speed, $\theta$ is the initial angle, measured in radians, and $g$ is the gravitational acceleration of earth.

Observe that, as a consequence of these equations, there are two times $t$ for which $y(t) = 0$.
One of them is $t = 0$, which is where the projectile starts its motion.
The other is $t = T$, where $T = 2v_0\sin(\theta)/g$ is the duration of the motion.
Accordingly, the distance this projectile will travel is $D = v_0^2\cos(\theta)\sin(\theta)/g$

```python
gravitational_acceleration = 9.8 # measured in meters per second per second
```

```python
import numpy as np # Needed to be able to have pi and sin

angle_radians = angle * np.pi / 180
duration = 2 * initial_velocity * np.sin(angle_radians) / gravitational_acceleration # measured in seconds
print(f"The duration of the trajectory was {duration} seconds")
```

```python
# List of times that we will poll for the location of the projectile

times = [0, 0.1, 0.2, 0.25, 0.33, 0.4, 0.45, 0.5, 0.65, 0.74, 0.81, 0.88, 0.92, 1.01, 1.02]
print("These are the times where we will learn where the projectile is, measured in seconds")
print(times)
```

```python
# List of x and y coordinates

positions_x = []
positions_y = []
for t in times:
  positions_x.append(initial_velocity * np.cos(angle_radians) * t)
  positions_y.append(initial_velocity * np.sin(angle_radians) * t - gravitational_acceleration * t**2 / 2)
print("These are the x")
print(positions_x)
print("and y")
print(positions_y)
print("coordinates of the projectile, measured in miles")
```

# Analyze our results with a plot

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.plot(positions_x, positions_y)
ax.set_xlabel(r"$x$(m)")
ax.set_ylabel(r"$y$(m)")
```
