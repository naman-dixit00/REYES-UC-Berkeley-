## Harmonic Oscillator

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import HTML
import matplotlib.animation as animation

t_max = 3

t = np.linspace(0, t_max, 1000)
x = np.cos(2*np.pi*t)

num_frames = 500

period_spring = 0.1
amplitude_spring = 0.03

fig, ax = plt.subplots(figsize = (15, 5))
ax.plot(t, x, c = "black", alpha = 0.5)
ax.plot([0, 0], [-1, 1], ls = "--", alpha = 0.5, c = "black")
ax.set_xlabel(r"$t/T$")
ax.set_ylabel(r"$x/A$")
spring, = ax.plot([], [])
particle = ax.scatter([], [])
particle_image = ax.scatter([], [], c = "blue")

def update(frame):

  t_particle = frame*3/num_frames

  spring_position = np.linspace(1, np.cos(2*np.pi*t_particle), 100)
  spring_figure = amplitude_spring*(np.sin(2*np.pi*spring_position/period_spring))

  particle.set_offsets(np.column_stack((0, np.cos(2*np.pi*t_particle))))
  spring.set_data(spring_figure, spring_position)

  particle_image.set_offsets(np.column_stack((t_particle, np.cos(2*np.pi*t_particle))))

anim = animation.FuncAnimation(fig, update, frames = num_frames, interval = 60000/num_frames)

anim.save('spring_motion.mp4', dpi=80, writer='ffmpeg')

plt.close(fig)

HTML(anim.to_html5_video())
```

```python
size_tangent = 0.25

fig, ax = plt.subplots(figsize = (15, 5))

ax.plot(t, x, c = "black", alpha =0.5)
ax.plot([0, 0], [-1, 1], ls = "--", alpha = 0.5, c = "black")

ax.set_xlabel(r"$t/T$")
ax.set_ylabel(r"$x/A$")

spring, = ax.plot([], [], c = "blue")
particle = ax.scatter([], [], c = "blue")
particle_image = ax.scatter([], [], c = "blue")
tangent, = ax.plot([], [], lw = 3, c = "orange")

def update(frame):

  t_particle = frame*3/num_frames

  spring_position = np.linspace(1, np.cos(2*np.pi*t_particle), 100)
  spring_figure = amplitude_spring*(np.sin(2*np.pi*spring_position/period_spring))

  particle.set_offsets(np.column_stack((0, np.cos(2*np.pi*t_particle))))
  spring.set_data(spring_figure, spring_position)

  particle_image.set_offsets(np.column_stack((t_particle, np.cos(2*np.pi*t_particle))))

  slope = -2*np.pi*np.sin(2*np.pi*t_particle)
  size_horizontal = np.cos(np.arctan(slope))*size_tangent
  tangent_times = np.linspace(t_particle - size_horizontal/2, t_particle + size_horizontal/2, 2)
  tangent_line = slope*(tangent_times - t_particle) + np.cos(2*np.pi*t_particle)
  tangent.set_data(tangent_times, tangent_line)

anim = animation.FuncAnimation(fig, update, frames = num_frames, interval = 60000/num_frames)

anim.save('spring_slope.mp4', dpi=80, writer='ffmpeg')

plt.close(fig)

HTML(anim.to_html5_video())
```

## Implementing Derivatives Numerically

### Derivatives of known functions

```python
import numpy as np
import matplotlib.pyplot as plt

def D(x, f, delta):

  """Defines the rate of change of a function f of x

  The final point is xf = x + delta
  The initial point is xi = x
  The difference between these two is delta"""

  return (f(x+delta)-f(x))/delta
```

#### Sine

```python
deltas = [10, 1, 0.1]

def f(x):
  return np.sin(x)

x = np.linspace(0, 6*np.pi, 10000)
y = f(x)
Dy = [D(x, f, delta) for delta in deltas]

fig, ax = plt.subplots(2, 1, sharex = True, figsize = (10,11))
ax[0].plot(x, y)
for i in range(len(deltas)):
  ax[1].plot(x, Dy[i], label = r"$\delta x = {}$".format(deltas[i]))
ax[1].plot(x, np.cos(x), label = r"$\cos(x)$")
ax[1].set_xlabel(r"$x$")
ax[0].set_ylabel(r"$\sin(x)$")
ax[1].set_ylabel(r"$\frac{d}{dx}\sin(x)$")
ax[1].legend()
```

#### Exponential

```python
deltas = [1, 0.1, 0.01]

def f(x):
  return np.exp(x)

x = np.linspace(0, 2, 10000)
y = f(x)
Dy = [D(x, f, delta) for delta in deltas]

fig, ax = plt.subplots(2, 1, sharex = True, sharey = True, figsize = (10,11))
ax[0].plot(x, y)
for i in range(len(deltas)):
  ax[1].plot(x, Dy[i], label = r"$\delta x = {}$".format(deltas[i]))
ax[1].plot(x, np.exp(x), label = r"$e^x$")
ax[1].set_xlabel(r"$x$")
ax[0].set_ylabel(r"$e^x$")
ax[1].set_ylabel(r"$\frac{d}{dx}e^x$")
ax[1].legend()
```

#### Logarithm

```python
deltas = [1, 0.1, 0.01, 0.001]

def f(x):
  return np.log(x)

x = np.linspace(0.5, 2, 10000)
y = f(x)
Dy = [D(x, f, delta) for delta in deltas]

fig, ax = plt.subplots(2, 1, sharex = True, figsize = (10,11))
ax[0].plot(x, y)
for i in range(len(deltas)):
  ax[1].plot(x, Dy[i], label = r"$\delta x = {}$".format(deltas[i]))
ax[1].plot(x, 1/x, label = r"$\frac{1}{x}$")
ax[1].set_xlabel(r"$x$")
ax[0].set_ylabel(r"$\ln(x)$")
ax[1].set_ylabel(r"$\frac{d}{dx}\ln(x)$")
ax[1].legend()
```

#### Hat Potential

```python
deltas = [1, 0.1, 0.01, 0.001]

def f(x):
  return (x**2 - 1)**2

x = np.linspace(-1.5, 1.5, 10000)
y = f(x)
Dy = [D(x, f, delta) for delta in deltas]

fig, ax = plt.subplots(2, 1, sharex = True, figsize = (10,10))
ax[0].plot(x, y)
for i in range(len(deltas)):
  ax[1].plot(x, Dy[i], label = r"$\delta x = {}$".format(deltas[i]))
ax[1].plot(x, 4*x*(x**2-1), label = r"$4x(x^2-1)$")
ax[1].set_xlabel(r"$x$")
ax[0].set_ylabel(r"$(x^2-1)^2$")
ax[1].set_ylabel(r"$\frac{d}{dx}(x^2-1)^2$")
ax[1].legend()
```

### Convergence

```python
x = 3

delta_deltas = 0.0001

def f(x):
  return np.sin(x)

deltas = np.arange(0.001, 1, delta_deltas)

fig, ax = plt.subplots()
ax.plot(deltas, D(x, f, deltas))
ax.set_xlabel(r"$\Delta x$")
ax.set_ylabel(r"$\left.\frac{{d}}{{dx}}\sin(x)\right|_{{x={}}}$".format(x))
```

```python
x = 3

delta_deltas = 0.0001

def D(x, f, delta):
  return (f(x + delta) - f(x - delta))/(2*delta)

def f(x):
  return np.sin(x)

deltas = np.arange(0.001, 1, delta_deltas)

fig, ax = plt.subplots(2, 1, sharex = True, figsize = (10,10))
ax[0].plot(deltas, D(x, f, deltas))
ax[1].plot(deltas, [(D(x, f, delta + delta_deltas)-D(x, f, delta))/delta_deltas for delta in deltas])
ax[1].set_xlabel(r"$\Delta x$")
ax[0].set_ylabel(r"$\left.\frac{{d}}{{dx}}\sin(x)\right|_{{x={}}}$".format(x))
ax[1].set_ylabel(r"$\frac{{d}}{{d\Delta x}}\left.\frac{{d}}{{dx}}\sin(x)\right|_{{x={}}}$".format(x))
```

```python
x = 0

delta_deltas = 0.0001

def D(x, f, delta):
  return (f(x + delta) - f(x - delta))/(2*delta)

def f(x):
  return np.exp(x)

deltas = np.arange(0.001, 1, delta_deltas)

fig, ax = plt.subplots(2, 1, sharex = True, figsize = (10,10))
ax[0].plot(deltas, D(x, f, deltas))
ax[1].plot(deltas, [(D(x, f, delta + delta_deltas)-D(x, f, delta))/delta_deltas for delta in deltas])
ax[1].set_xlabel(r"$\Delta x$")
ax[0].set_ylabel(r"$\left.\frac{{d}}{{dx}}e^x\right|_{{x={}}}$".format(x))
ax[1].set_ylabel(r"$\frac{{d}}{{d\Delta x}}\left.\frac{{d}}{{dx}}e^x\right|_{{x={}}}$".format(x))
```

## Applications

### Finding Minima

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1.5, 1.5, 100)

def f(x):
  return (x**2-1)**2+x/4

fig, ax = plt.subplots()
ax.plot(x, f(x))
extrema = np.array([-1.03, 0.0627, 0.967])
ax.scatter(extrema, f(extrema))
for xstar in extrema:
  horizontal = np.linspace(xstar - 0.2, xstar + 0.2, 2)
  vertical = np.repeat(f(xstar), 2)
  ax.plot(horizontal, vertical, c = "orange")
ax.set_xlabel(r"$x$")
ax.set_ylabel(r"$(x^2-1)^2+\frac{1}{4}x$")
```

```python
from IPython.display import HTML
import matplotlib.animation as animation

num_frames = 500

x = np.linspace(0, 6*np.pi, 100)
y = np.cos(x)
inty = np.sin(x)

fig, ax = plt.subplots(2, 1, sharex = True)

def update(frame):
  x_frame = frame * 6*np.pi/num_frames
  ax[0].clear()
  ax[1].clear()
  ax[0].plot(x, y, color = "#1f77b4")
  ax[0].set_ylabel(r"$\cos(x)$")
  ax[1].plot(x, inty, color = "#1f77b4")
  ax[1].set_ylabel(r"$\int_0^x\cos(\tilde{x})d\tilde{x}$")
  ax[1].set_xlabel(r"$x$")
  ax[1].scatter(x_frame, np.sin(x_frame), color = "#1f77b4")
  x_frame = np.linspace(0, x_frame, 100)
  ax[0].fill_between(x_frame, np.cos(x_frame), color = "#1f77b4", alpha = 0.3)

anim = animation.FuncAnimation(fig, update, frames = num_frames, interval = 60000/num_frames)

plt.close(fig)

anim.save('integration.mp4', dpi=80, writer='ffmpeg')

HTML(anim.to_html5_video())
```
